```python
#Question:1:--> Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog
# that overrides the speak() method to print "Bark!"?
# Answer-1
# Creating the parent class
class Animal:
    def speak(self):
        print("The animal makes a sound")  # Generic message for any animal

# Creating the child class that inherits from Animal
class Dog(Animal):
    def speak(self):
        print("Bark!")  # Specific message for Dog

# Creating objects and testing
cow = Animal()
cow.speak()      #This will give output: The animal makes a sound

rocky = Dog()
rocky.speak()    #This will give output: Bark!
```

```python
# Question:2:--->>>>Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle
# from it and implement the area() method in both.
# Answer2-->>>
from abc import ABC, abstractmethod

# Creating abstract class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass  # Only a declaration, no body

# Child class 1: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius  # Formula for area of circle

# Child class 2: Rectangle
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth  # Formula for area of rectangle

# Creating objects and calling area method
circle_obj = Circle(5)
print("Area of Circle:", circle_obj.area())   #This will give output: Area of Circle: 78.5

room = Rectangle(4, 6)
print("Area of Rectangle:", room.area())      #This will give output: Area of Rectangle: 24
```

```python
#Question-3--->>Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car
# and further derive a class ElectricCar that adds a battery attribute.
# Answer-3-->>>
# Base class
class Vehicle:
    def __init__(self, v_type):
        self.type = v_type  # Type of vehicle like 'Four Wheeler', 'Two Wheeler'

# First level derived class
class Car(Vehicle):
    def __init__(self, v_type, brand):
        super().__init__(v_type)  # Calling parent class constructor
        self.brand = brand        # Brand of the car like 'Maruti', 'Tata'

# Second level derived class
class ElectricCar(Car):
    def __init__(self, v_type, brand, battery_capacity):
        super().__init__(v_type, brand)       # Calling Car class constructor
        self.battery = battery_capacity       # Battery capacity in kWh

# Creating object of ElectricCar
neeraj_car = ElectricCar("Four Wheeler", "Mahindra", "30kWh")

# Accessing attributes
print("Vehicle Type:", neeraj_car.type)           #This will give output: Vehicle Type: Four Wheeler
print("Car Brand:", neeraj_car.brand)             #This will give output: Car Brand: Mahindra
print("Battery Capacity:", neeraj_car.battery)    #This will give output: Battery Capacity: 30kWh
```

```python
# Question--4--->>>Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classes
# Sparrow and Penguin that override the fly() method.
# Answer--4-->>
# Base class
class Bird:
    def fly(self):
        print("Some bird is flying...")  # General message for all birds

# Derived class 1
class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high in the sky!")  # Specific behavior of sparrow

# Derived class 2
class Penguin(Bird):
    def fly(self):
        print("Penguins can't fly, they swim instead!")  # Unique behavior of penguin

# Creating objects and demonstrating polymorphism
bird1 = Sparrow()
bird2 = Penguin()

# Same method name but different outputs depending on object
bird1.fly()   #This will give output: Sparrow flies high in the sky!
bird2.fly()   #This will give output: Penguins can't fly, they swim instead!
```

```python
# Question--5-->>>Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
# balance and methods to deposit, withdraw, and check balance
# Answer--5---->>>>
# Class to demonstrate encapsulation
class BankAccount:
    def __init__(self, account_holder):
        self.__balance = 0          # Private attribute
        self.holder = account_holder

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} deposited to {self.holder}'s account")  # Successful deposit
        else:
            print("Invalid deposit amount")  # Handling invalid input

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"{amount} withdrawn from {self.holder}'s account")  # Successful withdrawal
        else:
            print("Insufficient balance or invalid amount")  # Handling errors

    def check_balance(self):
        print(f"{self.holder}'s current balance is: {self.__balance}")  # Showing balance

# Creating object
rahul_acc = BankAccount("Rahul")

# Performing operations
rahul_acc.deposit(1000)       #This will give output: 1000 deposited to Rahul's account
rahul_acc.withdraw(300)       #This will give output: 300 withdrawn from Rahul's account
rahul_acc.check_balance()     #This will give output: Rahul's current balance is: 700
```

```python
# Question--6-->>>>Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar
# and Piano that implement their own version of play()?
# Answer-6->>>
# Base class
class Instrument:
    def play(self):
        print("Instrument is playing music")  # General message

# Derived class 1
class Guitar(Instrument):
    def play(self):
        print("Guitar is strumming chords")  # Specific implementation

# Derived class 2
class Piano(Instrument):
    def play(self):
        print("Piano is playing melodies")  # Specific implementation

# Function to demonstrate runtime polymorphism
def start_music(instrument_obj):
    instrument_obj.play()  # This will call the correct version based on object

# Creating objects
g = Guitar()
p = Piano()

# Passing objects to function
start_music(g)   #This will give output: Guitar is strumming chords
start_music(p)   #This will give output: Piano is playing melodies
```

```python
# Question---7-->>>> Create a class MathOperations with a class method add_numbers() to add two numbers and a static
# method subtract_numbers() to subtract two numbers?
# Answer--7--->>>>
class MathOperations:
    # Class method
    @classmethod
    def add_numbers(cls, a, b):
        return a + b  # Adds two numbers

    # Static method
    @staticmethod
    def subtract_numbers(x, y):
        return x - y  # Subtracts two numbers

# Using class method
result1 = MathOperations.add_numbers(15, 7)
print("Addition Result:", result1)   #This will give output: Addition Result: 22

# Using static method
result2 = MathOperations.subtract_numbers(20, 5)
print("Subtraction Result:", result2)  #This will give output: Subtraction Result: 15
```

```python
# Question:8--->>> Implement a class Person with a class method to count the total number of persons created?
# Answer-8--->>>
class Person:
    count = 0  # Class variable to keep track of total persons

    def __init__(self, name):
        self.name = name
        Person.count += 1  # Increment count when new object is created

    @classmethod
    def total_persons(cls):
        print("Total persons created:", cls.count)  # Shows total number

# Creating objects
p1 = Person("Anjali")
p2 = Person("Ravi")
p3 = Person("Sundar")

# Calling class method to see total persons
Person.total_persons()   #This will give output: Total persons created: 3
```

```python
# Question:9--->>>> Write a class Fraction with attributes numerator and denominator. Override the str method to display the
# fraction as "numerator/denominator"?
# Answer:--->>>>
class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"  # Custom string format

# Creating object
f1 = Fraction(3, 4)
print(f1)   #This will give output: 3/4

f2 = Fraction(7, 2)
print(f2)   #This will give output: 7/2
```

```python
# Question-10->>>Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
# vectors?
# Answer:--->>>
class Vector:
    def __init__(self, x, y):
        self.x = x  # X-component of vector
        self.y = y  # Y-component of vector

    def __add__(self, other):
        # Overloading + operator to add two vectors
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        # Custom string to show vector nicely
        return f"({self.x}, {self.y})"

# Creating two vector objects
v1 = Vector(3, 5)
v2 = Vector(4, 2)

# Adding the vectors using overloaded +
result = v1 + v2
print("Resultant Vector:", result)   #This will give output: Resultant Vector: (7, 7)
```

```python
# Question-12-->>> Implement a class Student with attributes name and grades. Create a method average_grade() to compute
# the average of the grades.
# Answer-12->>>>
class Student:
    def __init__(self, name, grades):
        self.name = name           # Student's name
        self.grades = grades       # List of grades

    def average_grade(self):
        if len(self.grades) == 0:
            return 0  # To handle case if no grades are there
        return sum(self.grades) / len(self.grades)  # Average calculation

# Creating object with grades
rajesh = Student("Rajesh", [85, 90, 78, 92])

# Printing average
print("Average Grade of", rajesh.name, "is:", rajesh.average_grade())
#This will give output: Average Grade of Rajesh is: 86.25
```

```python
# Question-13--->>>>Create a class Rectangle with methods set_dimensions() to set the dimensions and area() to calculate the
# area?
# Answer-13->>>>
class Rectangle:
    def set_dimensions(self, length, breadth):
        self.length = length      # Setting length of rectangle
        self.breadth = breadth    # Setting breadth of rectangle

    def area(self):
        return self.length * self.breadth  # Area formula

# Creating object and setting dimensions
my_plot = Rectangle()
my_plot.set_dimensions(8, 5)

# Calculating and printing area
print("Area of rectangle is:", my_plot.area())
#This will give output: Area of rectangle is: 40
```

```python
# Question-14--->>>>Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
# and hourly rate. Create a derived class Manager that adds a bonus to the salary?
# Answer-14--->>>>
# Base class
class Employee:
    def __init__(self, name, hours, rate):
        self.name = name
        self.hours = hours              # Total hours worked
        self.rate = rate                # Hourly rate

    def calculate_salary(self):
        return self.hours * self.rate   # Basic salary calculation

# Derived class
class Manager(Employee):
    def __init__(self, name, hours, rate, bonus):
        super().__init__(name, hours, rate)
        self.bonus = bonus              # Extra bonus for managers

    def calculate_salary(self):
        base = super().calculate_salary()  # Get base salary from Employee
        return base + self.bonus           # Add bonus for manager

# Creating objects
emp1 = Employee("Kiran", 40, 200)
print("Employee Salary:", emp1.calculate_salary())
#This will give output: Employee Salary: 8000

mgr1 = Manager("Sheetal", 40, 200, 3000)
print("Manager Salary:", mgr1.calculate_salary())
#This will give output: Manager Salary: 11000
```

```python
# Question--15--->>>>Create a class Product with attributes name, price, and quantity. Implement a method total_price() that
# calculates the total price of the product
# Answer-15--->>>>
class Product:
    def __init__(self, name, price, quantity):
        self.name = name           # Name of the product
        self.price = price         # Price per item
        self.quantity = quantity   # Total quantity bought

    def total_price(self):
        return self.price * self.quantity  # Total cost formula

# Creating object
item = Product("Parle-G Biscuits", 10, 5)

# Calculating total price
print("Total price for", item.name, "is:", item.total_price())
#This will give output: Total price for Parle-G Biscuits is: 50
```

```python
# Question:16--->>>>Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that
# implement the sound() method
# Answer:16--->>>>
from abc import ABC, abstractmethod

# Abstract base class
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass  # Only a declaration, no implementation

# Derived class 1
class Cow(Animal):
    def sound(self):
        print("Cow says: Hamba Hamba")  # Cow's sound

# Derived class 2
class Sheep(Animal):
    def sound(self):
        print("Sheep says: Bhaa Bhaa")  # Sheep's sound

# Creating objects and calling sound
gaumata = Cow()
gaumata.sound()    #This will give output: Cow says: Hamba Hamba

meenu = Sheep()
meenu.sound()      #This will give output: Sheep says: Bhaa Bhaa
```

```python
# Question:17:--->>> Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that
# returns a formatted string with the book's details?
# Answer-17--->>>>
class Book:
    def __init__(self, title, author, year_published):
        self.title = title                    # Book title
        self.author = author                  # Author name
        self.year_published = year_published  # Year of publishing

    def get_book_info(self):
        return f"'{self.title}' by {self.author}, published in {self.year_published}"  # Formatted info

# Creating object
novel = Book("The Secrets of Sambar", "Raghav Deshmukh", 2021)

# Getting book info
print(novel.get_book_info())
#This will give output: 'The Secrets of Sambar' by Raghav Deshmukh, published in 2021
```

```python
# Question-18--->>> Create a class House with attributes address and price. Create a derived class Mansion that adds an
# attribute number_of_rooms.
# Answer:--->>>
# Base class
class House:
    def __init__(self, address, price):
        self.address = address    # Address of the house
        self.price = price        # Price of the house

# Derived class
class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)       # Initialize base class attributes
        self.number_of_rooms = number_of_rooms  # Extra attribute for mansion

# Creating object of derived class
big_house = Mansion("Juhu Beach Road, Mumbai", 50000000, 12)

# Printing details
print("Address:", big_house.address)                #This will give output: Address: Juhu Beach Road, Mumbai
print("Price:", big_house.price)                    #This will give output: Price: 50000000
print("Number of Rooms:", big_house.number_of_rooms)  #This will give output: Number of Rooms: 12
```
